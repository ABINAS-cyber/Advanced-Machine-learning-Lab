### Mutual Information

```python
import math
import numpy as np
import pandas as pd

data = np.array([[8, 2], [5, 3], [1, 1]])

rows = ["High", "medium", "low"]
columns = ["pass", "fail"]

df = pd.DataFrame(data, index=rows, columns=columns)
print(df)

total = data.sum()
print("\n total observations =", total)

row_totals = data.sum(axis=1)
column_totals = data.sum(axis=0)

print("\n Row totals =", row_totals)
print("column totals =", column_totals)
print("MARGINAL PROBABILITIES")
print("\n P(Result):")

for i in range(len(rows)):
  probability = row_totals[i] / total
  print(f"P({rows[i]}) = {row_totals[i]}/{total} = {probability:.3f}")

print("\n P(Study level):")

for j in range(len(columns)):
  probability = column_totals[j] / total
  print(f"P({columns[j]}) = {column_totals[j]}/{total} = {probability:.3f}")

MI = 0
for i in range(len(rows)):
  for j in range(len(columns)):
    p_xy = data[i][j] / total
    p_x = row_totals[i] / total
    p_y = column_totals[j] / total
    ratio = p_xy / (p_x * p_y)
    contribution = p_xy * math.log2(ratio)
    MI = MI + contribution

    print(f"\n cell : ({rows[i]}, {columns[j]})")
    print(f" count = {data[i][j]}")
    print(f" P(x,y) = {data[i][j]}/{total} = {p_xy:.3f}")
    print(f" P(x) = {row_totals[i]}/{total} = {p_x:.3f}")
    print(f" P(y) = {column_totals[j]}/{total} = {p_y:.3f}")
    print(f" Ratio = P(x, y)/[P(x)*P(y)] = {ratio:.3f}")
    print(
        f" Contribution = {p_xy:.3f} * log2({ratio:.3f}) = {contribution:.3f}"
    )

print(f"\nTotal mutual Information (MI) = {MI:.4f}")
```
