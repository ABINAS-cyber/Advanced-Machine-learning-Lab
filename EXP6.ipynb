### Correlation based feature selection (CFS)

```python
import pandas as pd

students = ["A", "B", "C", "D", "E", "F"]
study_hours = [2, 4, 6, 8, 10, 12]
marks = [50, 60, 80, 70, 90, 100]

df = pd.DataFrame(
    {"Student": students, "Study Hours (X)": study_hours, "Marks (Y)": marks}
)

print("Original Data")
print(df)

df["Rank X"] = (
    df["Study Hours (X)"].rank(method="average", ascending=True).astype(int)
)
print(df)

df["Rank Y"] = (
    df["Marks (Y)"].rank(method="average", ascending=True).astype(int)
)
df["d"] = df["Rank X"] - df["Rank Y"]
print(df)

df["d2"] = df["d"] ** 2
print(df)

sum_d2 = df["d2"].sum()
print("\n sum of d2 =", sum_d2)

n = len(df)
print("number of students (n) :", n)

rho = 1 - (6 * sum_d2) / (n * (n**2 - 1))
print(f"Spearman's Rank Correlation Coefficient (p) = {rho:.4f}")

if rho > 0.9:
  print("\n Interpretation : very strong positive correlation")
elif rho > 0.7:
  print("\n Interpretation : strong positive correlation")
elif rho > 0.3:
  print("\n Interpretation : moderate positive correlation")
elif rho > 0:
  print("\n Interpretation : weak positive correlation")
elif rho == 0:
  print("\n Interpretation : no correlation")
else:
  print("\n Interpretation : negative correlation")
```
