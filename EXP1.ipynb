### Logistic Regression

```python
import numpy as np
import math

x1 = np.array([3, 1, 2, 4, 3])
x2 = np.array([6, 3, 4, 7, 5])
y = np.array([1, 0, 0, 1, 1])

X = np.column_stack((np.ones(5), x1, x2))
B = np.linalg.inv(X.T @ X) @ X.T @ y

b0, b1, b2 = B
print("Co-efficient for b1:", b1)
print("Co-efficient for b2:", b2)
print("Intercept (b0):", b0)
```

```python
def sigmoid(x):
    return 1 / (1 + math.exp(-x))

z = X @ B
print("Linear combination (z):")
print(z)

y_prob = np.array([sigmoid(val) for val in z])
print("\nPredicted probabilities (sigmoid output):")
print(y_prob)

y_pred = (y_prob >= 0.5).astype(int)
print("\nPredicted classes:")
print(y_pred)
```

```python
TP = np.sum((y == 1) & (y_pred == 1))
TN = np.sum((y == 0) & (y_pred == 0))
FP = np.sum((y == 0) & (y_pred == 1))
FN = np.sum((y == 1) & (y_pred == 0))

print("True Positives (TP):", TP)
print("True Negatives (TN):", TN)
print("False Positives (FP):", FP)
print("False Negatives (FN):", FN)

accuracy = (TP + TN) / (TP + TN + FP + FN) if (TP + TN + FP + FN) != 0 else 0
precision = TP / (TP + FP) if (TP + FP) != 0 else 0
recall = TP / (TP + FN) if (TP + FN) != 0 else 0
f1_score = (2 * precision * recall) / (precision + recall) if (precision + recall) != 0 else 0

print("\nAccuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1_score)
```
