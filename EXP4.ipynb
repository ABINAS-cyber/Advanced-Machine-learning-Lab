### AVNOVA and CHI-SQUARE

```python
import numpy as np
from scipy.stats import f

samsung = np.array([36000, 35000, 31000, 42000, 38000])
google = np.array([44000, 45000, 51000, 50000, 52000])
sony = np.array([16000, 15000, 18000, 24000, 22000])

mean_samsung = np.mean(samsung)
mean_google = np.mean(google)
mean_sony = np.mean(sony)

print("mean samsung :", mean_samsung)
print("mean google :", mean_google)
print("mean sony :", mean_sony)

all_data = np.concatenate([samsung, google, sony])
grand_mean = np.mean(all_data)
print("Grand mean =", grand_mean)

n_samsung = len(samsung)
n_google = len(google)
n_sony = len(sony)

SSB = (
    n_samsung * (mean_samsung - grand_mean) ** 2
    + n_google * (mean_google - grand_mean) ** 2
    + n_sony * (mean_sony - grand_mean) ** 2
)
print("SSB =", SSB)

ssw_samsung = np.sum((samsung - mean_samsung) ** 2)
ssw_google = np.sum((google - mean_google) ** 2)
ssw_sony = np.sum((sony - mean_sony) ** 2)

SSW = ssw_samsung + ssw_google + ssw_sony
print("SSW samsung:", ssw_samsung)
print("SSW google:", ssw_google)
print("SSW sony:", ssw_sony)
print("total SSW =", SSW)

k = 3
N = len(all_data)
df_between = k - 1
df_within = N - k
df_total = N - 1

print("df Between =", df_between)
print("df within =", df_within)
print("df total =", df_total)

MSB = SSB / df_between
MSW = SSW / df_within

print("MS Between =", MSB)
print("MS within =", MSW)

F_calculated = MSB / MSW
print("F calculated =", F_calculated)

alpha = 0.05
F_critical = f.ppf(1 - alpha, df_between, df_within)

print("Alpha =", alpha)
print("F critical =", F_critical)

if F_calculated > F_critical:
  print("Reject H0")
  print("There is a significant difference among the group means")
else:
  print("Fail to Reject H0")
  print("There is No significant diff among the group means")
```

### CHI-square

```python
import numpy as np
from scipy.stats import chi2

observed = np.array([[45, 5], [15, 35]])

row_totals = observed.sum(axis=1)

print("OBSERVED FREQUENCY TABLE")
print("-" * 50)
print("                    Pass      Fail")
print("high Attendance =", observed[0])
print("Low Attendance =", observed[1])

print("\nRow totals")
print("-" * 50)
print("High Attendance =", row_totals[0])
print("low Attendance =", row_totals[1])

column_totals = observed.sum(axis=0)
print("\nColumn Totals")
print("-" * 50)
print("Pass =", column_totals[0])
print("Fail =", column_totals[1])

grand_total = observed.sum()
print("\nGRAND TOTAL")
print("-" * 50)
print("Grand Total =", grand_total)

expected = np.zeros_like(observed, dtype=float)
for i in range(observed.shape[0]):
  for j in range(observed.shape[1]):
    expected[i, j] = (row_totals[i] * column_totals[j]) / grand_total

print(expected)

difference = observed - expected
print(difference)

chi_square_components = (difference**2) / expected
print("\nCHI-square contributions of each cell")
print(chi_square_components)

chi_square = chi_square_components.sum()
print("\nCHI-square calculation")
print("Chi-square =", chi_square)

rows, columns = observed.shape
df = (rows - 1) * (columns - 1)
print("\nDEGREES OF FREEDOM")
print("Rows =", rows)
print("Columns =", columns)
print("df =", df)

alpha = 0.05
critical_value = chi2.ppf(1 - alpha, df)
print("\ncritical value")
print("Alpha =", alpha)
print("Critical chi-square =", critical_value)

if chi_square > critical_value:
  print("Reject H0")
  print("Attendance and Result are significantly associated")
else:
  print("Fail to Reject H0")
  print("Attendance and Result are independent")
```
