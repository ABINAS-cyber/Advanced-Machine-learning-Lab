### L1 and L2 regularization

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Lasso, LinearRegression, Ridge
from sklearn.metrics import mean_squared_error

X = np.array([
    [1, 2, 10],
    [2, 1, 12],
    [3, 3, 9],
    [4, 5, 11],
    [5, 4, 13],
    [6, 6, 8],
    [7, 7, 7],
    [8, 8, 14],
])

y = np.array([6, 7, 12, 18, 20, 24, 28, 32])

df = pd.DataFrame(X, columns=["X1", "X2", "X3"])
df["y"] = y

print("Dataset:\n", df)
```

```python
# 1. Without Regularization
lr = LinearRegression()
lr.fit(X, y)
y_pred_lr = lr.predict(X)

print("Linear Regression (No Regularization)")
print("Coefficients:", lr.coef_)
print("Intercept:", lr.intercept_)
print("MSE:", mean_squared_error(y, y_pred_lr))
```

```python
# 2. Ridge Regression (L2)
ridge = Ridge(alpha=1.0)
# alpha = regularization strength
ridge.fit(X, y)
y_pred_ridge = ridge.predict(X)

print("Ridge Regression (L2 Regularization)")
print("Coefficients:", ridge.coef_)
print("Intercept:", ridge.intercept_)
print("MSE:", mean_squared_error(y, y_pred_ridge))
```

```python
# 3. Lasso Regression (L1)
lasso = Lasso(alpha=0.5)
lasso.fit(X, y)
y_pred_lasso = lasso.predict(X)

print("Lasso Regression (L1 Regularization)")
print("Co-efficients:", lasso.coef_)
print("MSE:", mean_squared_error(y, y_pred_lasso))
```
