```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b", 
    api_key="your - Api - Key"
)
```

```python
with open("./blood_work.txt","r") as f:
  blood_report = f.read()
print(blood_report)
```

```python
extraction_prompt  = f"""
you are a medical data extraction assistant.open

From the blood report below , extract all test values
and classify each one as HIGH , LOW or NORMAL
based on the reference range provided in the report.

Format your response as :
- Test Name : value | status : HIGH/LOW/NORMAL | Reference : range

Blood Report:
{blood_report}
"""

extraction_response = llm.invoke(extraction_prompt)

print(extraction_response.content)
```

```python
diet_prompt = f"""
you are a clinical nutritionist specializing in indian dietary habits.

based on the blood work analysis below,write:
1. A short health summary in 4-5 lines explaning
the patient's condition in simple language
2. A short , practical indian diet plan having only two sections
(1)Foods to avoid (2)Foods to eat more of .
Do not include any other sections in diet plan.

Blood Work Analysis:
{extraction_response}
"""

diet_response = llm.invoke(diet_prompt)

print(diet_response.content)
```
