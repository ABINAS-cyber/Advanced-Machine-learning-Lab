```python
from google import genai
from PIL import Image

# Initialize the Google GenAI client
client = genai.Client(api_key="YOUR_API_KEY")

# Load your image
image = Image.open("sample_image.jpg")

# Pass both text and image to the multi-modal agent
prompt = "Analyze this image and tell me what you see."

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=[prompt, image]
)

print("Agent Response:\n", response.text)
```

```python
from google import genai

client = genai.Client(api_key="YOUR_API_KEY")

def ask_agent(role_prompt, task):
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=[f"{role_prompt}\n\nTask: {task}"]
    )
    return response.text

# 1. The Planner Agent breaks down the task
task = "Write a python script to scrape a website and save to CSV."
planner_prompt = "You are a planning agent. Break this task down into 2 clear steps."
plan = ask_agent(planner_prompt, task)

print("--- PLANNER OUTPUT ---")
print(plan)

# 2. The Worker Agent executes the plan
worker_prompt = "You are a coding agent. Write the code based strictly on this plan."
code_result = ask_agent(worker_prompt, plan)

print("\n--- WORKER OUTPUT ---")
print(code_result)
```
